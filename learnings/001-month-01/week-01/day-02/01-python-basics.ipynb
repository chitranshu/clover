### Control Structures
Python provides control structures such as if statements, for loops, and while loops to control the flow of your program. These structures allow you to execute code based on certain conditions or to repeat code multiple times.

````python
# Example of Control Structures

# If statement
age = 18
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")

# For loop
print("For loop:")
for i in range(5):
    print(i)

# While loop
print("While loop:")
count = 0
while count < 5:
    print(count)
    count += 1
````

### Operators
Python supports various operators for performing operations on variables and values. These include arithmetic operators, comparison operators, logical operators, and assignment operators.

#### Arithmetic Operators
- `+` (Addition)
- `-` (Subtraction)
- `*` (Multiplication)
- `/` (Division)
- `//` (Floor Division)
- `%` (Modulus)
- `**` (Exponentiation)
#### Comparison Operators
- `==` (Equal to)
- `!=` (Not equal to)
- `>` (Greater than)
- `<` (Less than)
- `>=` (Greater than or equal to)
- `<=` (Less than or equal to)
#### Logical Operators
- `and` (Logical AND)
- `or` (Logical OR)
- `not` (Logical NOT)
#### Assignment Operators
- `=` (Assignment)
- `+=` (Addition Assignment)
- `-=` (Subtraction Assignment)
- `*=` (Multiplication Assignment)
- `/=` (Division Assignment)
- `//=` (Floor Division Assignment)
- `%=` (Modulus Assignment)
- `**=` (Exponentiation Assignment)
#### Bitwise Operators
- `&` (Bitwise AND)
- `|` (Bitwise OR)
- `^` (Bitwise XOR)
- `~` (Bitwise NOT)
- `<<` (Left Shift)
- `>>` (Right Shift)

````python
# Operators
# Arithmetic operators
a = 10
b = 5
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

# Comparision operators
print("Equal to:", a == b)
print("Not equal to:", a != b)
print("Greater than:", a > b)
print("Less than:", a < b)
print("Greater than or equal to:", a >= b)
print("Less than or equal to:", a <= b)

# Logical operators
print("Logical AND:", a > 0 and b > 0)
print("Logical OR:", a > 0 or b < 0)
print("Logical NOT:", not (a > 0))

# Assignment operators
c = 10
c += 5  # c = c + 5
print("After += 5:", c)
c -= 3  # c = c - 3
print("After -= 3:", c)
c *= 2  # c = c * 2
print("After *= 2:", c)
c /= 4  # c = c / 4
print("After /= 4:", c)
c //= 2  # c = c // 2
print("After //= 2:", c)
c %= 3  # c = c % 3
print("After %= 3:", c)
c **= 2  # c = c ** 2
print("After **= 2:", c)

# Bitwise operators
x = 5  # 0101 in binary
y = 3  # 0011 in binary
print("Bitwise AND:", x & y)  # 0001 in binary
print("Bitwise OR:", x | y)   # 0111 in binary
print("Bitwise XOR:", x ^ y)  # 0110 in binary
print("Bitwise NOT:", ~x)     # 1010 in binary (two's complement)
print("Left Shift:", x << 1)  # 1010 in binary
print("Right Shift:", x >> 1) # 0010 in binary
````

### Strings 

Strings are sequences of characters enclosed in quotes. You can use single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """) to define strings in Python.

#### Example of a string
```python
greeting = "Hello, World!"
print(greeting)
```



````python
# Example of Strings
greeting = "Hello"
name = "Alice"
message = greeting + ", " + name + "!"
print(message)

greeting = 'Hello'
name = 'Alice'
message = greeting + ', ' + name + '!'
print(message)

greeting = '''Hello'''
name = '''Alice'''
message = greeting + ', ' + name + '!'
print(message)

greeting = """Hello"""
name = """Alice"""
message = greeting + ', ' + name + '!'
print(message)
````

### String Operators
- Concatenation: `+` operator can be used to concatenate two strings.
- Repetition: `*` operator can be used to repeat a string a specified number of times.
- Membership: `in` operator can be used to check if a substring exists within a string.

````python
# Example of string operators
str1 = "Hello"
str2 = "World"
# Concatenation
result = str1 + " " + str2
print("Concatenation:", result)
# Repetition
repeated = str1 * 3
print("Repetition:", repeated)
# Membership
print("Membership:", "lo" in str1)
print("Membership:", "Hi" in str1)
````

### String comparison
You can compare strings using comparison operators. For example:
```python

print(str1 == str2)  # Output: False
print(str1 != str2)  # Output: True
print(str1 < str2)   # Output: True (because "Hello" comes before "World" in lexicographical order)

```

### String Methods
Python provides various built-in methods for string manipulation. Some common string methods include:
- `lower()`: Converts a string to lowercase.
- `upper()`: Converts a string to uppercase.
- `strip()`: Removes leading and trailing whitespace from a string.
- `split()`: Splits a string into a list of substrings based on a specified delimiter.
- `join()`: Joins a list of strings into a single string with a specified delimiter.
- `replace()`: Replaces occurrences of a specified substring with another substring.
- `find()`: Returns the index of the first occurrence of a specified substring, or -1 if the substring is not found.
- `count()`: Returns the number of occurrences of a specified substring in the string.


````python
# String methods
text = "Hello, World!"
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Title Case:", text.title())
print("Strip:", text.strip("!"))
print("Replace:", text.replace("World", "Python"))
print("Split:", text.split(", "))
print("Find:", text.find("World"))
print("Count:", text.count("o"))
````

### List operations
Python provides various operations that can be performed on lists, such as:
- Adding elements: You can add elements to a list using the `append()` method or the `extend()` method.
- Removing elements: You can remove elements from a list using the `remove()` method or the `pop()` method.
- Slicing: You can slice a list to get a subset of its elements using the slicing syntax `list[start:stop:step]`.
- Iterating: You can iterate over the elements of a list using a for loop or a while loop.

````python
# List operations
# Adding elements
my_list = [1, 2, 3, "Hello", 4.5]
my_list.append("New Element")
print("After append:", my_list)
my_list.extend([6, 7, 8])
print("After extend:", my_list)
# Removing elements
my_list.remove("Hello")
print("After remove:", my_list)
popped_element = my_list.pop()
print("Popped element:", popped_element)
print("After pop:", my_list)
# Slicing
sliced_list = my_list[1:4]
print("Sliced list:", sliced_list)
# Iterating
print("Iterating over list:")
for item in my_list:
    print(item)
````

# Data types and converstion
Python has several built-in data types, including:
- `int`: Integer numbers (e.g., 1, 2, 3)
- `float`: Floating-point numbers (e.g., 1.0, 2.5)
- `str`: Strings (e.g., "Hello", "World")
- `bool`: Boolean values (e.g., True, False)
- `list`: Lists (e.g., [1, 2, 3], ["Hello", "World"])
- `tuple`: Tuples (e.g., (1, 2, 3), ("Hello", "World"))
- `dict`: Dictionaries (e.g., {"name": "Alice", "age": 30})
- `set`: Sets (e.g., {1, 2, 3})

````python
# Example of data types and conversion
# Data types
integer_value = 10
float_value = 3.14
string_value = "Hello"
boolean_value = True
print("Integer:", integer_value)
print("Float:", float_value)
print("String:", string_value)
print("Boolean:", boolean_value)

# Type conversion
num_str = "123"
num_int = int(num_str)
print("String to Integer:", num_int)

float_str = "3.14"
float_num = float(float_str)
print("String to Float:", float_num)
int_to_str = str(num_int)
print("Integer to String:", int_to_str)
bool_to_str = str(boolean_value)
print("Boolean to String:", bool_to_str)
````
